# Calibration exploration
Run the fit script first with a sourced, versioned dataset. This notebook reads its output without changing calibration membership.

```python
import json
from pathlib import Path
result = json.loads(Path('fit.json').read_text())
print('Dataset:', result['dataset_version'])
print('Calibration IDs:', result['calibration_ids'])
print('Preformation:', result['preformation'])
```

```python
import matplotlib.pyplot as plt
for split in ('calibration', 'validation'):
    rows = [r for r in result['cases'] if r['split'] == split]
    plt.scatter([r['log10_half_life_s'] for r in rows], [r['predicted_log10_half_life_s'] for r in rows], label=split)
plt.xlabel('Observed log10 half-life (s)')
plt.ylabel('Predicted log10 half-life (s)')
plt.legend();
```
